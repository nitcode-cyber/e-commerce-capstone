2. Duplicated order‑item rows?

```python
dupes = df.duplicated(subset=['order_id','order_item_id']).sum()
print("\nDuplicate order‑item rows:", dupes)
```

3.Purchse date range

```python
print("Order purchase from", df['order_purchase_timestamp'].min(),
"to", df['order_purchase_timestamp'].max())
```

4. Delivery before purchase? (negative delivery times)

```python
df['delivery_time'] = (df['order_delivered_customer_date'] -
df['order_purchase_timestamp']).dt.days
neg = (df['delivery_time'] < 0).sum()
print("Orders with negative delivery_time:", neg)
```

Ensures your timestamp parsing worked

Catches impossible records (deliveries before purchase)

How many orders per customer?

```python
items_per_order = df.groupby ('order_id')['order_item_id'].nunique(). describe()
print("\nItems per order:\n", items_per_order)
```

7.Correlation & Heatmap (Before Scaling)

```python
numeric = ['price','freight_value','delivery_time']
corr = df[numeric].corr()
sns.heatmap(corr, annot=True, fmt=".2f")
plt.title("Numeric Correlations")
plt.show()
```

Spot multicollinearity (e.g. price ↔ freight)
